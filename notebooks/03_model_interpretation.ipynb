# Customer Churn Prediction — Model Interpretation

This notebook visualizes the Logistic Regression coefficients from the trained churn model and documents the observed business signals.

**Important:** coefficients indicate model association/predictive signal, not causation.

```python
import pandas as pd
import matplotlib.pyplot as plt

feature_importance = pd.DataFrame({
    'Feature': [
        'Tenure', 'Contract: Two year', 'Monthly charges', 'Internet: DSL',
        'Contract: Month-to-month', 'Internet: Fiber optic', 'Total charges',
        'Streaming TV: Yes', 'Streaming Movies: Yes', 'Payment: Electronic check',
        'Online Security: No', 'Tech Support: No'
    ],
    'Coefficient': [
        -1.250374, -0.777701, -0.621137, -0.588834,
        0.696066, 0.660906, 0.607433, 0.252081, 0.244732,
        0.229096, 0.199371, 0.176853
    ]
})

feature_importance['Direction'] = feature_importance['Coefficient'].apply(
    lambda x: 'Higher churn association' if x > 0 else 'Lower churn association'
)
feature_importance.sort_values('Coefficient', ascending=False)
```

```python
plot_df = feature_importance.sort_values('Coefficient')

plt.figure(figsize=(10, 7))
plt.barh(plot_df['Feature'], plot_df['Coefficient'])
plt.axvline(0, linewidth=1)
plt.xlabel('Logistic Regression coefficient')
plt.ylabel('Feature')
plt.title('Top Logistic Regression Churn Signals')
plt.tight_layout()
plt.savefig('../visualizations/logistic_regression_coefficients.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Observed model signals

### Higher churn association
- Month-to-month contract (0.696)
- Fiber optic internet (0.661)
- Total charges (0.607)
- Streaming TV: Yes (0.252)
- Streaming Movies: Yes (0.245)
- Electronic check payment (0.229)
- Online Security: No (0.199)
- Tech Support: No (0.177)

### Lower churn association
- Tenure (-1.250)
- Two-year contract (-0.778)
- Monthly charges (-0.621)
- DSL internet (-0.589)

The monthly-charge coefficient is counterintuitive and should not be interpreted as evidence that higher prices reduce churn. Correlation among predictors can affect individual regression coefficients.

Repeated 'No internet service' coefficients were not treated as separate business discoveries because they overlap with the same underlying service condition.
